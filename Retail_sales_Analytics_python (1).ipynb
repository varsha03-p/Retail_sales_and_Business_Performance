```python
import matplotlib.pyplot as plt

category_sales = df.groupby('Product_Category')['Sales_Amount'].sum()

category_sales.plot(kind='bar', figsize=(8,5), title='Sales by Product Category')

plt.xlabel('Product Category')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
region_sales = df.groupby('Region')['Sales_Amount'].sum().sort_values(ascending=False)

print(region_sales)
```

```python
region_mapping = {
    'North East': 'Northeast',
    'NRW': 'North Rhine-Westphalia',
    'British columbia': 'British Columbia'
}

df['Region'] = df['Region'].replace(region_mapping)

region_sales = df.groupby('Region')['Sales_Amount'].sum().sort_values(ascending=False)

print(region_sales)
```

```python
region_sales.plot(
    kind='bar',
    figsize=(12, 6),
    title='Region-wise Sales Analysis'
)

plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
monthly_sales = df.groupby('Year_Month')['Sales_Amount'].sum().sort_index()

print(monthly_sales)
```

```python
monthly_sales.plot(
    kind='line',
    figsize=(14, 6),
    marker='o',
    title='Monthly Sales Trend (2022–2025)'
)

plt.xlabel('Year-Month')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=90)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
top_products = df.groupby('Product_Name')['Sales_Amount'].sum().sort_values(ascending=False).head(10)

print(top_products)
```

```python
top_products.plot(
    kind='bar',
    figsize=(12,6),
    title='Top 10 Products by Sales'
)

plt.xlabel('Product Name')
plt.ylabel('Sales Amount')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
top_profit = df.groupby('Product_Name')['Profit'].sum().sort_values(ascending=False).head(10)

print(top_profit)
```
